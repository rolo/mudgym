# Quickstart

Create a MudGym env and send an action.

```python
from mudgym import make_env
from mudgym.notebooks import show_ansi, show_table
```

```python
env = make_env(observation="parsed", actions="text", render_mode="ansi")
observation, info = env.reset(seed=7)
show_ansi(info["render_bytes"])
```

```python
info["render_bytes"]
```

```python
show_ansi(info["render_bytes"])
```

```python
info["raw_bytes"]
```

```python
observation
```

```python
env.action_space
```

```python
action = "howl"
observation, reward, terminated, truncated, info = env.step(action)
```

```python
info["render_bytes"]
```

```python
show_ansi(info["render_bytes"])
```

```python
show_table(
    [
        {
            "Action": action,
            "Room": observation["room_name"],
            "Reward": reward,
            "Terminated": terminated,
            "Truncated": truncated,
        }
    ]
)
```

```python
show_ansi(info["raw_bytes"])
```
